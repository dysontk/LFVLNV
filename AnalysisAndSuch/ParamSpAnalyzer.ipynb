```python
def getPreExistingFolders(start_d):
    fromls = AM.run_command('ls ' + start_d)
    # print(fromls.split)
    # print([str(b) for b in fromls.split()])
    # if fromls[0] != 'L':
    #     # then we are not looking good
    #     print("no processes at all")
    #     return []
    # else:
    runlist = [r[5:-6] for r in fromls.split()]
    # print([r[5:-6] for r in runlist])
    # print(runlist)
    return runlist
    
def get_runs(sd, foldername):
    # n_runfolders = 0
    runlist = [r[5:-6] for r in AM.run_command('ls ' + sd + foldername).split()]
    n_runfolders = len(runlist)
    n_runs = 0
    if not n_runfolders:
        return 0
    else:
        events = []
        for run in runlist: 
            with open(sd+foldername+ run + '/delphes_events.dat', 'r') as fl:
                events = fl.readlines()
                events = int(events[0])
            if events>=3000:
                n_runs += 1
    return n_runs 

def checkPreexistingProc(L, g):
    newFolderName = PSAnal.fileNameMaker(L, g)
    preexistingFolders = getPreExistingFolders(start_dir)
    alreadyExists = False
    for folder in preexistingFolders:
        if newFolderName == folder:
            alreadyExists = True
    
    if not alreadyExists:
        return 0
    else:
        return get_runs(start_dir, newFolderName+'/')
```

```python
import numpy as np
def incrementParam(info):
    newinfo = info
    newinfo.update({'current': round(info['current']+ info['delta'], ndigits=5)})
    return newinfo

def set_start(dict):
    # print(dict)
    newdict = dict
    newdict.update({'current': dict['bounds'][0]})
    return newdict

```

```python
def make_safe_grid():
    LI = {'bounds':(1000, 5e3), # GeV
                'delta': 500/2}
    gI = {'bounds':(0.1, 1.1), 
                'delta': 0.2/2}
    LI = set_start(LI)
    gI = set_start(gI)
    outlist = []
    while LI['current'] <= LI['bounds'][1]:
        
        # print("Lambda: ", LambdaInfo['current'])
        geffInfo = set_start(gI)
        templist = []
        while gI['current'] <= gI['bounds'][1]:
            outlist.append((LI['current'], gI['current']))
            gI = incrementParam(gI)
        # outlist.append(templist)
        LI = incrementParam(LI)
    return np.array(outlist)

def in_safe_grid(LI, GI, s_grid):
    print(f"checking whether to delete {LI['current'], GI['current']} runs")
    in_grid = False
    # print(s_grid)
    for pair in s_grid:
        # print(pair)
        # print(np.array((LI['current'], GI['current'])))
        # print((np.array((LI['current'], GI['current'])) == pair).all())
        if (np.array((LI['current'], GI['current'])) == pair).all():
            # print("hi")
            in_grid = True
    return in_grid

def remove_runs(LI, gi):
    AM.run_command(f"rm -vr /work/pi_mjrm_umass_edu/LNV_collider/Generated/Signal/{PSAnal.fileNameMaker(LI, gi)}")


LambdaInfo = set_start(LambdaInfo)
geffInfo = set_start(geffInfo)


```
